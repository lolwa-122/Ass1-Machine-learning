```python
df = df[df['Age'] > 0]
```

```python
df = df[df['Age'] <= 100]
```

```python
df = df.drop_duplicates()
```

```python
df = df[df['WaitingDays'] >= 0]
```

```python
df = df.copy()
```

```python
df['No-show'] = df['No-show'].map({'No': 0, 'Yes': 1})
```

```python
print("After Cleaning:")
print(df.shape)
print(df['Age'].min())
print(df.duplicated().sum())
```

```python
###Data Cleaning

The dataset was cleaned to improve data quality and ensure accurate analysis.

- Invalid age values were removed by filtering out unrealistic values (age greater than 100).
- Duplicate rows were identified and removed to avoid redundant data.
- A new feature "WaitingDays" was cleaned by removing negative values, which represent incorrect records.
- The "No-show" column was converted into numerical format for easier analysis.

### Before vs After Cleaning

- Before cleaning: 110,526 rows  
- After cleaning: 106,987 rows  
- Total rows removed: 3,539  

The minimum age changed from 0 to 1, indicating that non-informative or edge-case values were removed.

Overall, the dataset became cleaner, more consistent, and more suitable for analysis.
```

```python
##The cleaning process significantly improved the dataset by removing noisy and inconsistent data, resulting in more reliable and meaningful insights.
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20)
plt.title("Distribution of Age")
plt.show()
```

```python
- This plot shows the distribution of patient ages.
- Most patients are concentrated in specific age ranges.
- It helps identify whether the data is skewed or contains unusual values.
```

```python
sns.countplot(x='Gender', hue='No-show', data=df)
plt.title("Gender vs No-show")
plt.show()
```

```python
- This plot shows the relationship between gender and appointment attendance.
- Both males and females show similar patterns.
- There is no strong difference in attendance based on gender.
```

```python
sns.countplot(x='No-show', data=df)
plt.title("Show vs No-show")
plt.show()
```

```python
- This plot compares patients who attended their appointments versus those who did not.
- The majority of patients attended their appointments.
- A noticeable portion did not show up.
```

```python
sns.countplot(x='SMS_received', hue='No-show', data=df)
plt.title("SMS Received vs No-show")
plt.show()
```

```python
- This plot shows the effect of SMS reminders on attendance.
- Patients who received SMS tend to show slightly better attendance.
- SMS reminders may help reduce no-show rates.
```

```python
sns.boxplot(x='No-show', y='Age', data=df)
plt.title("Age vs No-show")
plt.show()
```

```python
- This plot compares age distribution between patients who showed up and those who did not.
- Younger patients may have slightly higher no-show rates.
- Age might influence attendance behavior.
```

```python
sns.countplot(x='Diabetes', hue='No-show', data=df)
plt.title("Diabetes vs No-show")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Age'], fill=True)
plt.title("Age Density Distribution")
plt.show()
```

```python
- This plot shows the density distribution of patient ages.
- It provides a smoother view compared to the histogram.
- Most patients are concentrated in middle age ranges.
```

```python
plt.figure(figsize=(10,6))
correlation = df.corr(numeric_only=True)
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python


- This heatmap shows the relationships between numerical variables in the dataset.
- Most correlations are weak, indicating that no single factor strongly determines attendance.
- Age shows slight correlation with some medical conditions such as hypertension.
- The dataset does not have strong linear relationships between variables.
```
